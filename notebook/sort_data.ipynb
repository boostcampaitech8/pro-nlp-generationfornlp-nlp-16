### ensemble 결과 중 차이가 있거나 vote_count가 문제들로만 test_data 재구성

```python
import pandas as pd

soft_df = pd.read_csv("../data/output_soft.csv")
hard_df = pd.read_csv("../data/output_hard.csv")

# soft_df, hard_df 가 동일한 index를 가진다고 가정
diff_mask = soft_df["answer"] != hard_df["answer"]
diff_rows = soft_df[diff_mask]

test_df = pd.read_csv("../data/test.csv")

filtered_df = test_df[test_df["id"].isin(diff_rows["id"])]

filtered_df.to_csv(
    "../data/soft_hard_diff_data.csv",
    index=False,
    encoding="utf-8-sig"
)
```

```python
len(filtered_df)
```

```python
ensemble_df = pd.read_csv("../data/ensemble_analysis.csv")

target_ids = ensemble_df.loc[ensemble_df["vote_count"] <= 3, "id"]

filtered_df2 = test_df[test_df["id"].isin(target_ids)]
filtered_df2.to_csv(
    "ensemble_diff_data.csv",
    index=False,
    encoding="utf-8-sig"
)
```

### MoA로 재추론 문제들 정답 교체

```python
soft_df = pd.read_csv("../data/output_soft.csv")
exaone_37 = pd.read_csv("../data/exaone_diff_37.csv")


# id 기준으로 exaone_37의 answer를 soft_df에 붙임
soft_df_updated = soft_df.merge(
    exaone_37[["id", "answer"]],
    on="id",
    how="left",
    suffixes=("", "_exaone")
)

# answer가 다를 때만 exaone_37 값으로 교체
mask = (
    soft_df_updated["answer_exaone"].notna() &
    (soft_df_updated["answer"] != soft_df_updated["answer_exaone"])
)

soft_df_updated.loc[mask, "answer"] = soft_df_updated.loc[mask, "answer_exaone"]

# 보조 컬럼 제거
soft_df_updated = soft_df_updated.drop(columns=["answer_exaone"])
```

```python
soft_df_updated.to_csv(
    "../diff37_changed_submission.csv",
    index=False,
    encoding="utf-8-sig"
)
```
