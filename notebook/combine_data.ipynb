```python
import pandas as pd
from pathlib import Path
import ast
```

```python
# 파일 경로 설정
train_csv = "../data/train.csv"
augmented_csv = "../data/train_combined_augmented.csv"
workbook_csv = "../data/aihub_workbook.csv"

# 데이터 로드
df_train = pd.read_csv(train_csv)
df_augmented = pd.read_csv(augmented_csv)
df_workbook = pd.read_csv(workbook_csv)

print(f"train.csv: {len(df_train):,}개 row")
print(f"train_combined_augmented.csv: {len(df_augmented):,}개 row")
print(f"aihub_workbook.csv: {len(df_workbook):,}개 row")
```

```python
df_combined = pd.concat([df_train, df_augmented, df_workbook], ignore_index=True)

print(f"\n통합 완료: 총 {len(df_combined):,}개 row")
```

```python
# 저장 경로
output_path = "../data/train_all_combined.csv"

# CSV 저장
df_combined.to_csv(output_path, index=False, encoding='utf-8')

print(f"\n통합 데이터 저장 완료: {output_path}")
```
